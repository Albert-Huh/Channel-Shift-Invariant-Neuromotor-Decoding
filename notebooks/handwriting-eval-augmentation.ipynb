## Train and Eval

```python
# training & eval helpers (uses your loaded `config`)


pl.seed_everything(42, workers=True)

def _cfg_with_data_location(cfg: DictConfig, data_dir: str) -> DictConfig:
    c = copy.deepcopy(cfg)
    c["data_module"]["data_location"] = os.path.expanduser(data_dir)
    if "from_csv" in c["data_module"]["data_split"]["_target_"]:
        c["data_module"]["data_split"]["csv_filename"] = os.path.join(
            os.path.expanduser(data_dir), f"{TASK_NAME}_corpus.csv"
        )
    return c

def _set_aug(cfg: DictConfig, train=None, val=None, test=None) -> DictConfig:
    """
    Set channel_aug per split. Pass a small dict or None.
    Example train: {"rotation":{"enable":True,"random_k":True,"k_range":[-8,8]}, "permutation":{"enable":False}}
    """
    c = copy.deepcopy(cfg)
    c.setdefault("data_module", {}).setdefault("channel_aug", {})
    if train is not None: c["data_module"]["channel_aug"]["train"] = train
    if val   is not None: c["data_module"]["channel_aug"]["val"] = val
    if test  is not None: c["data_module"]["channel_aug"]["test"] = test
    return c

def _build_trainer(run_name: str, max_epochs: int = None, precision: str = "16-mixed", accum: int = 4) -> Trainer:
    accel = "gpu" if torch.cuda.is_available() else "cpu"
    devices = 1
    log_dir = Path("./outputs") / f"{TASK_NAME}_{run_name}"
    tb = TensorBoardLogger(save_dir=str(log_dir), name="tb")

    monitor, mode = "val/CER", "min"  # keep same monitor key you had
    callbacks = [
        ModelCheckpoint(monitor=monitor, mode=mode, save_top_k=1, save_last=True, filename="{epoch}-{"+monitor+":.4f}"),
        EarlyStopping(monitor=monitor, mode=mode, patience=20, check_finite=False),
    ]

    return Trainer(
        accelerator=accel, devices=devices,
        precision=precision,
        accumulate_grad_batches=accum,
        max_epochs=max_epochs or 120,
        log_every_n_steps=50, enable_progress_bar=True,
        callbacks=callbacks, logger=tb
    )


def train_once(run_name: str, cfg: DictConfig):
    """
    Trains a model from scratch using lightning_module config (same as pretrained),
    returns (best_ckpt_path, log_dir)
    """
    cfg = _cfg_with_data_location(cfg, EMG_DATA_DIR)
    # Instantiate module from config (NOT from checkpoint)
    module = instantiate(cfg["lightning_module"])
    datamodule = instantiate(cfg["data_module"])

    trainer = _build_trainer(run_name)
    trainer.fit(module, datamodule=datamodule)

    # Get best checkpoint
    ckpt_cb = [cb for cb in trainer.callbacks if isinstance(cb, ModelCheckpoint)][0]
    best = ckpt_cb.best_model_path or ckpt_cb.last_model_path
    log_dir = Path(trainer.logger.save_dir) / trainer.logger.name / f"version_{trainer.logger.version}"
    return best, str(log_dir)

# ---- evaluation helpers (clean vs augmented test) ----

# Fallback collate wrapper if your installed package doesn't accept `channel_aug` in DataModule
from torch.utils.data import DataLoader
from generic_neuromotor_interface.utils import handwriting_collate

def _make_aug_collate(rotation=None, permutation=None, seed=42):
    import torch
    gen = torch.Generator().manual_seed(int(seed))
    def _collate(samples):
        batch = handwriting_collate(samples)
        x = batch["emg"]   # (N,T,C)
        C = int(x.shape[-1])
        if rotation and rotation.get("enable", False):
            if rotation.get("random_k", False):
                lo, hi = rotation.get("k_range", [-4, 4])
                k = int(torch.randint(low=int(lo), high=int(hi)+1, size=(1,), generator=gen))
            else:
                k = int(rotation.get("k", 0))
            k = ((k % C) + C) % C
            if k > C // 2: k -= C
            x = torch.roll(x, shifts=k, dims=-1)
        if permutation and permutation.get("enable", False):
            if permutation.get("mode","random") == "random":
                perm = torch.randperm(C, generator=gen, device=x.device)
            else:
                perm = torch.as_tensor(permutation["perm"], dtype=torch.long, device=x.device)
            x = x.index_select(-1, perm)
        batch["emg"] = x
        return batch
    return _collate

def eval_on_test(model_or_ckpt, cfg: DictConfig, test_aug=None):
    """
    Evaluate on test. If DataModule supports channel_aug, set it there.
    Otherwise, wrap the test dataloader with collate aug.
    """
    cfg = _cfg_with_data_location(cfg, EMG_DATA_DIR)

    # Try DataModule channel_aug first
    dm_cfg = copy.deepcopy(cfg["data_module"])
    try:
        if test_aug is not None:
            dm_cfg.setdefault("channel_aug", {})
            dm_cfg["channel_aug"]["test"] = test_aug
        datamodule = instantiate(dm_cfg)
        use_fallback = False
    except TypeError:
        # Installed package doesn't accept channel_aug -> fallback to collate wrapper
        datamodule = instantiate(cfg["data_module"])
        use_fallback = True

    # Load model from ckpt or use instance
    if isinstance(model_or_ckpt, (str, Path)) and os.path.exists(str(model_or_ckpt)):
        model = HandwritingModule.load_from_checkpoint(str(model_or_ckpt), map_location="cpu")
    else:
        model = model_or_ckpt

    trainer = _build_trainer(run_name="eval", max_epochs=1)

    if not use_fallback or test_aug is None:
        # plain path
        results = trainer.test(model=model, datamodule=datamodule)
    else:
        # fallback: build augmented test dataloader only
        datamodule.setup("test")
        base_loader = datamodule.test_dataloader()
        aug_collate = _make_aug_collate(rotation=test_aug.get("rotation"),
                                        permutation=test_aug.get("permutation"))
        aug_loader = DataLoader(
            datamodule.test_dataset,
            batch_size=base_loader.batch_size,
            shuffle=False,
            num_workers=base_loader.num_workers,
            pin_memory=True,
            collate_fn=aug_collate,
        )
        results = trainer.test(model=model, dataloaders=aug_loader)
    return results[0] if results else {}
```

```python
# ==== Launch training runs ====
# base copies of the pretrained config (architecture/optim match pretrained)
base_cfg = copy.deepcopy(config)
base_cfg["data_module"]["batch_size"] = 10
base_cfg["data_module"]["num_workers"] = 10

# A) No augmentation anywhere (train/val/test)
noaug_cfg = _set_aug(base_cfg,
    train={"rotation":{"enable":False}, "permutation":{"enable":False}},
    val  ={"rotation":{"enable":False}, "permutation":{"enable":False}},
    test ={"rotation":{"enable":False}, "permutation":{"enable":False}},
)
ckpt_noaug, log_noaug = train_once("noaug", noaug_cfg)
print("NoAug best ckpt:", ckpt_noaug)

# B) Train 
# with rotation aug; val/test clean (for fair model selection & headline)
trainaug_cfg = _set_aug(base_cfg,
    train={"rotation":{"enable":True, "random_k":True, "k_range":[-8,8]}, "permutation":{"enable":False}},
    val  ={"rotation":{"enable":False}, "permutation":{"enable":False}},
    test ={"rotation":{"enable":False}, "permutation":{"enable":False}},
)
ckpt_trainaug, log_trainaug = train_once("trainaug", trainaug_cfg)
print("TrainAug best ckpt:", ckpt_trainaug)
```

```python
# ==== Comparison matrix ====
from pprint import pprint

# 1) Pretrained vs New-NoAug on Test
print("\n[1] Test NoAug: Pretrained vs New-NoAug")
res_pretrained_clean = eval_on_test(model_ckpt_path, base_cfg, test_aug=None)
res_noaug_clean      = eval_on_test(ckpt_noaug,        base_cfg, test_aug=None)
pprint({"pretrained": res_pretrained_clean, "new_noaug": res_noaug_clean})

# 2) Pretrained vs New-NoAug vs New-TrainAug on:
# (a) Test NoAUG
print("\n[2a] Test NoAug: Pretrained vs New-NoAug vs New-TrainAug")
res_trainaug_clean = eval_on_test(ckpt_trainaug, base_cfg, test_aug=None)
pprint({"pretrained": res_pretrained_clean, "new_noaug": res_noaug_clean, "new_trainaug": res_trainaug_clean})

# (b) Test AUGMENTED (rotation only by default; enable permutation for extreme test)
test_aug_rotation = {"rotation":{"enable":True, "random_k":True, "k_range":[-8,8]},
                     "permutation":{"enable":False}}
print("\n[2b] Test AUG (rotation): Pretrained vs New-NoAug vs New-TrainAug")
res_pretrained_aug = eval_on_test(model_ckpt_path, base_cfg, test_aug=test_aug_rotation)
res_noaug_aug      = eval_on_test(ckpt_noaug,      base_cfg, test_aug=test_aug_rotation)
res_trainaug_aug   = eval_on_test(ckpt_trainaug,   base_cfg, test_aug=test_aug_rotation)
pprint({"pretrained": res_pretrained_aug, "new_noaug": res_noaug_aug, "new_trainaug": res_trainaug_aug})

# (c) (Optional) Test AUG with rotation + permutation
test_aug_perm = {"rotation":{"enable":True, "random_k":True, "k_range":[-8,8]},
                 "permutation":{"enable":True, "mode":"random"}}
print("\n[2c] Test AUG (rotation+permutation): Pretrained vs New-NoAug vs New-TrainAug")
res_pretrained_perm = eval_on_test(model_ckpt_path, base_cfg, test_aug=test_aug_perm)
res_noaug_perm      = eval_on_test(ckpt_noaug,      base_cfg, test_aug=test_aug_perm)
res_trainaug_perm   = eval_on_test(ckpt_trainaug,   base_cfg, test_aug=test_aug_perm)
pprint({"pretrained": res_pretrained_perm, "new_noaug": res_noaug_perm, "new_trainaug": res_trainaug_perm})
```

## Run inference on one prompt

```python
"""Grab one test prompt"""

test_dataset = make_handwriting_dataset(
    dataset_names=["handwriting_user_001_dataset_000"],
    data_location=datamodule.data_location,
    transform=datamodule.transform,
    padding=datamodule.padding,
    emg_augmentation=None,
    concatenate_prompts=False,
    min_duration_s=0.0,
)

sample = test_dataset[55]  # an arbitrary prompt from this dataset
```

```python
"""Run inference"""

model.eval()

# unpack sample
emg = sample["emg"]
labels = sample["prompts"]

# compute model outputs
with torch.no_grad():
    emissions, _slice = model(emg.T.unsqueeze(0))

    # compute greedy decode outputs
    predictions = model.decoder.decode_batch(
        emissions=emissions.movedim(0, 1).numpy(),
        emission_lengths=model.network.compute_time_downsampling(
            emg_lengths=torch.as_tensor([len(emg)]), slc=_slice
        )
    )

predictions = torch.as_tensor(predictions[0])

# convert predictions and labels to characters
predictions = model.decoder._charset.labels_to_str(predictions)
labels = model.decoder._charset.labels_to_str(labels)
```

```python
"""Evaluate CER on this prompt"""

metric = CharacterErrorRates()
metric.update(
    prediction=predictions,
    target=labels,
)
aggregate_metrics = metric.compute()

print("CER of above prompt decode:", aggregate_metrics["CER"])
```

```python
"""Print predictions and target"""

print(
    f"Prediction: \t {predictions} \n"
    f"Target: \t {labels}"
)
```

## Test set evaluation: baseline (no aug) vs with channel augmentation

We reuse the pretrained model and run `trainer.test` twice:
1) baseline (no aug)
2) test-time channel rotation (and optionally rotation+permutation)


```python
from copy import deepcopy
import os
from hydra.utils import instantiate

# 0) Make sure data_location is correct in any cloned config
def _ensure_data_location(cfg):
    cfg = deepcopy(cfg)
    cfg["data_module"]["data_location"] = os.path.expanduser(EMG_DATA_DIR)
    return cfg
trainer = Trainer(accelerator="gpu")
```

```python
# 1) Baseline (no augmentation)
cfg_base = _ensure_data_location(config)
# (No channel_aug keys means your DataModule uses defaults -> no aug)
datamodule_base = instantiate(cfg_base["data_module"])
print("Running test WITHOUT augmentation...")
test_results_base = trainer.test(model=model, datamodule=datamodule_base)
print(test_results_base)
```
