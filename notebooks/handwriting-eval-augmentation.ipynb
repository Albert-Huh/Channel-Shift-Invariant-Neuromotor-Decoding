## Load model config

```python
"""Load model config"""

config_path = os.path.join(os.path.expanduser(MODELS_DIR), TASK_NAME, "model_config.yaml")
config = OmegaConf.load(config_path)
```

## Load model checkpoint

```python
"""Load model checkpoint"""
from generic_neuromotor_interface.lightning import HandwritingModule



model_ckpt_path = os.path.join(
    os.path.expanduser(MODELS_DIR),
    TASK_NAME,
    "model_checkpoint.ckpt"
)

model = HandwritingModule.load_from_checkpoint(
    model_ckpt_path,
    map_location=torch.device("cpu"),
)
# model = instantiate(config.lightning_module)
# model = model.load_from_checkpoint(
#     model_ckpt_path,
#     map_location=torch.device("cpu"),
# )
```

## Instantiate data module

```python
"""Assemble the data module"""

# Update DataModule config with data path
config["data_module"]["data_location"] = os.path.expanduser(EMG_DATA_DIR)
if "from_csv" in config["data_module"]["data_split"]["_target_"]:
    config["data_module"]["data_split"]["csv_filename"] = os.path.join(
        os.path.expanduser(EMG_DATA_DIR),
        f"{TASK_NAME}_corpus.csv"
    )

datamodule = instantiate(config["data_module"])
```

## Run inference on one prompt

```python
"""Grab one test prompt"""

test_dataset = make_handwriting_dataset(
    dataset_names=["handwriting_user_001_dataset_000"],
    data_location=datamodule.data_location,
    transform=datamodule.transform,
    padding=datamodule.padding,
    emg_augmentation=None,
    concatenate_prompts=False,
    min_duration_s=0.0,
)

sample = test_dataset[55]  # an arbitrary prompt from this dataset
```

```python
"""Run inference"""

model.eval()

# unpack sample
emg = sample["emg"]
labels = sample["prompts"]

# compute model outputs
with torch.no_grad():
    emissions, _slice = model(emg.T.unsqueeze(0))

    # compute greedy decode outputs
    predictions = model.decoder.decode_batch(
        emissions=emissions.movedim(0, 1).numpy(),
        emission_lengths=model.network.compute_time_downsampling(
            emg_lengths=torch.as_tensor([len(emg)]), slc=_slice
        )
    )

predictions = torch.as_tensor(predictions[0])

# convert predictions and labels to characters
predictions = model.decoder._charset.labels_to_str(predictions)
labels = model.decoder._charset.labels_to_str(labels)
```

```python
"""Evaluate CER on this prompt"""

metric = CharacterErrorRates()
metric.update(
    prediction=predictions,
    target=labels,
)
aggregate_metrics = metric.compute()

print("CER of above prompt decode:", aggregate_metrics["CER"])
```

```python
"""Print predictions and target"""

print(
    f"Prediction: \t {predictions} \n"
    f"Target: \t {labels}"
)
```

## Test set evaluation: baseline (no aug) vs with channel augmentation

We reuse the pretrained model and run `trainer.test` twice:
1) baseline (no aug)
2) test-time channel rotation (and optionally rotation+permutation)


```python
from copy import deepcopy
import os
from hydra.utils import instantiate

# 0) Make sure data_location is correct in any cloned config
def _ensure_data_location(cfg):
    cfg = deepcopy(cfg)
    cfg["data_module"]["data_location"] = os.path.expanduser(EMG_DATA_DIR)
    return cfg
trainer = Trainer(accelerator="gpu")
```

```python
# 1) Baseline (no augmentation)
cfg_base = _ensure_data_location(config)
# (No channel_aug keys means your DataModule uses defaults -> no aug)
datamodule_base = instantiate(cfg_base["data_module"])
print("Running test WITHOUT augmentation...")
test_results_base = trainer.test(model=model, datamodule=datamodule_base)
print(test_results_base)
```

```python
# 2) Test-time rotation augmentation (only on test split)
cfg_rot = _ensure_data_location(config)
cfg_rot["data_module"]["channel_aug"] = {
    "train": {"rotation": {"enable": False}, "permutation": {"enable": False}},
    "val":   {"rotation": {"enable": False}, "permutation": {"enable": False}},
    "test":  {"rotation": {"enable": True, "random_k": False, "k": 7},
              "permutation": {"enable": False}},
}
datamodule_rot = instantiate(cfg_rot["data_module"])
print("Running test WITH rotation augmentation...")
test_results_rot = trainer.test(model=model, datamodule=datamodule_rot)
print(test_results_rot)
```

```python
# 3) (Optional) Test-time rotation + permutation
cfg_perm = _ensure_data_location(config)
cfg_perm["data_module"]["channel_aug"] = {
    "train": {"rotation": {"enable": False}, "permutation": {"enable": False}},
    "val":   {"rotation": {"enable": False}, "permutation": {"enable": False}},
    "test":  {"rotation": {"enable": True, "random_k": True, "k_range": [-8, 8]},
              "permutation": {"enable": True, "mode": "random"}},
}
datamodule_perm = instantiate(cfg_perm["data_module"])
print("Running test WITH rotation + permutation augmentation...")
test_results_perm = trainer.test(model=model, datamodule=datamodule_perm)
print(test_results_perm)
```
